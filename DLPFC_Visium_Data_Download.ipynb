```python
# ============================================================
# STEP 1: Mount Google Drive
# ============================================================
from google.colab import drive
drive.mount('/content/drive')

import requests, os

# ============================================================
# STEP 2: Fetch file list from Figshare
# ============================================================
article_id = "28200299"
url = f"https://api.figshare.com/v2/articles/{article_id}/files"
response = requests.get(url)
files = response.json()

print("Available files:")
for f in files:
    print(f['name'], '---', f['download_url'])

# ============================================================
# STEP 3: Download only the 5 needed samples
# ============================================================
target_samples = ["151671", "151507", "151508", "151509", "151510"]

save_path = "/content/drive/MyDrive/visium_data/"
os.makedirs(save_path, exist_ok=True)

for f in files:
    name = f['name']
    # Only download files that match our target sample IDs
    if any(sample in name for sample in target_samples):
        dl_url = f['download_url']
        out_file = os.path.join(save_path, name)

        if os.path.exists(out_file):
            print(f"  ⏭ Already exists, skipping: {name}")
            continue

        print(f"Downloading {name}...")
        r = requests.get(dl_url, stream=True)
        with open(out_file, 'wb') as out:
            for chunk in r.iter_content(chunk_size=8192):
                out.write(chunk)
        print(f"  ✓ Done: {name}")

print("\n✅ All target samples downloaded!")
```
